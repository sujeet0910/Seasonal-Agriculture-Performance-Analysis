## Phase 3: Data Cleaning
First, let's check if our dataset has any duplicate rows.

```python
# Identifying duplicates
duplicates = df.duplicated().sum()
print(f"Total Duplicate Rows Found: {duplicates}")

df = df.drop_duplicates()
print(f"Duplicates handled. Current Data Shape: {df.shape}")
```

Now, let's find out exactly which columns contain missing (null) values and how many are missing.

```python
print(df.isnull().sum())
```

Fixing Rainfall Data: Since rainfall varies heavily by season, we will fill the missing 'Rainfall_mm' values using the average rainfall of their respective seasons, rather than the overall average.

```python

df['Rainfall_mm'] = df['Rainfall_mm'].fillna(df.groupby('Season')['Rainfall_mm'].transform('mean'))
print("Missing Rainfall values fixed.")
```

Fixing Soil Moisture Data: For soil moisture, we will use the median value to fill the missing spots, as median is less affected by extreme outliers.

```python

df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(df['Soil_Moisture_pct'].median())
print("Missing Soil Moisture values fixed.")
```

Fixing Yield Data: Crop yield is defined as Total Production divided by Total Farm Area. We will calculate and fill the missing yield values using this exact formula.

```python

df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(df['Production_Tonnes'] / df['Farm_Area_Hectares'])
print("Missing Yield values fixed.")
```

Final check to ensure our dataset is 100% clean and has zero missing values.

```python
print("Missing values after cleaning:")
print(df.isnull().sum().sum())
```

## Phase 8: Final Project Summary

### **8 Meaningful Insights from the Analysis**
1. **Top Performing Season:** The Kharif season consistently outperforms other seasons, yielding the highest crop production (Tonnes/Ha).
2. **Profitability Risk:** The Zaid season carries severe financial risks, often resulting in negative average profits for farmers.
3. **Water Dependency:** Zaid season crops consume significantly more water per hectare due to high average temperatures and lower natural rainfall.
4. **Natural Advantage:** Kharif benefits from an average of over 850mm of rainfall, drastically reducing the dependency on artificial irrigation and lowering total costs.
5. **Irrigation Inefficiency:** The traditional "Flood" irrigation method is the least efficient, producing the lowest yield per 1000m³ of water used.
6. **Modern Irrigation Benefits:** "Drip" and "Rainfed" methods show significantly higher water efficiency metrics, proving better for sustainable farming.
7. **Pest Vulnerability:** Disease and pest risks are heavily season-dependent, strongly correlating with changes in humidity and temperature.
8. **Economic Correlation:** There is a highly positive correlation between Total Cost and Revenue, but high revenue does not always translate to high profit (especially in Zaid), highlighting the impact of input costs.

### **Limitations of the Analysis**
*   **Time Span:** The dataset focuses on seasonal variations but lacks year-over-year data, meaning we cannot predict long-term climate change impacts.
*   **Imputation Constraints:** Missing rainfall values were filled using seasonal averages, which might mask localized micro-climatic variations in specific districts.
*   **Market Dynamics:** The analysis assumes a stable market price as given in the dataset, which in reality fluctuates daily based on supply, demand, and government policies.

### **Evidence-Based Recommendations**
1. **Season-Specific Planning:** Stakeholders should limit water-intensive crops in the Zaid season. Farmers should be encouraged to plant drought-resistant or short-duration crops to prevent financial losses.
2. **Irrigation Upgrades:** Since Flood irrigation is widely used but inefficient, agricultural departments should provide subsidies for Drip and Sprinkler irrigation systems to optimize water resource usage.
3. **Risk Mitigation:** For the Rabi and Zaid seasons, specialized pest control and soil moisture retention techniques must be proactively implemented to secure yields.

### **Final Conclusion**
This complete agricultural data analysis successfully reveals that farming performance is deeply interconnected with seasonal variations. The Kharif season remains the economic backbone for these farms, while the Zaid season requires major strategic overhauls in resource management. By shifting from traditional flood irrigation to efficient methods and optimizing crop choices per season, farmers can significantly improve both their water sustainability and financial profit.
